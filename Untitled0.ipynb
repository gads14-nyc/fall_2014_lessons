```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
data= pd.DataFrame({'Qu1':[1,3,4,3,4],
                    'Qu2':[2,3,1,2,3],
                    'Qu3':[6,9,0,4,3],
                    'Qu4':[8,4,5,3,4],
                    'Qu5':[1,5,2,4,4]})
```

```python
fig, axes = plt.subplots(2,2)
```

```python
k=0
predictors=[]
for i in data.columns:
    predictors.append(i)
for i in range(2):
    for j in range(2):
        axes[i,j].scatter(data[predictors[k]],data['Qu5'])
        k+=1
```

```python
plt.show()
```
